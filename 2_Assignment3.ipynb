# **Remove Punctution**

```python
import string
exclude = string.punctuation
def remove_punctution(text):
  return text.translate(str.maketrans("","",exclude))
```

# **Removing Stop Words**

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
# stop_words = set(stopwords.words('english'))
def remove_stopwords(text):
  new_text = []
  for word in text.split():
    if word in stopwords.words("english"):
      new_text.append("")
    else:
      new_text.append(word)
  x = new_text[:]
  new_text.clear()
  return " ".join(x)
```

```python
top_movies_df['overview'] = top_movies_df['overview'].apply(remove_stopwords)
```

# **Stemming**

```python
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
def stem_words(lst):
  return " ".join([ps.stem(str(word)) for word in lst])
```

```python
top_movies_df['overview'].apply(stem_words).head(2)
```

# **Lemmatization**

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
wnl = WordNetLemmatizer()

def wordnlemmatizer(lst):
  return " ".join([wnl.lemmatize(str(word),pos='v') for word in lst])
```

```python
top_movies_df['overview'].apply(wordnlemmatizer)
```
