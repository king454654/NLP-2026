```python
exclude = string.punctuation
```

```python
def remove_punctuation(text):
  for char in exclude:
    text = text.replace(char,"")
  return text
```

```python
def chat_conversion(text):
  new_text = []
  for w in text.split():
    if w.upper() in chat_words:
      new_text.append(chat_words[w.upper()])
    else:
      new_text.append(w)
  return " ".join(new_text)
```

```python
def remove_stopwords(text):
  new_text = []
  for word in text.split():
    if word in stopwords.words("english"):
      new_text.append("")
    else:
      new_text.append(word)
  x = new_text[:]
  new_text.clear()
  return " ".join(x)
```

```python
remove_stopwords("Hello I am Indrajit")
```

```python
# handling emojis

# Source - https://stackoverflow.com/a/33417311
# Posted by jfs, modified by community. See post 'Timeline' for change history
# Retrieved 2026-02-24, License - CC BY-SA 3.0

#!/usr/bin/env python
import re

def remove_emoji(text):
  emoji_pattern = re.compile("["
          u"\U0001F600-\U0001F64F"  # emoticons
          u"\U0001F300-\U0001F5FF"  # symbols & pictographs
          u"\U0001F680-\U0001F6FF"  # transport & map symbols
          u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                            "]+", flags=re.UNICODE)
  return emoji_pattern.sub(r'', text) # no emoji
```

```python
remove_emoji("Hello how are u?😁")
```

# Stemming

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
def stem_words(text):
  return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample = "walk walks walking walked"
stem_words(sample)
```

# Lemmatization

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
wnl = WordNetLemmatizer()

sentence = "He was runing and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations = "?:!.,;"
sentence_words = nltk.word_tokenize(sentence)

for word in sentence_words:
  if word in punctuations:
    sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word","Lemma"))
for word in sentence_words:
  print("{0:20}{1:20}".format(word,wnl.lemmatize(word,pos='v')))
```
